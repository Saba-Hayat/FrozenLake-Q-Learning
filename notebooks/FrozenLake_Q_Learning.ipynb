# FrozenLake-v1 Q-Learning Agent
Executed project notebook. This notebook loads the artifacts generated by the training script and presents evaluation results.

```python
from pathlib import Path
import json
import numpy as np
from IPython.display import Image, display
ROOT = Path.cwd()
with open(ROOT / 'outputs/evaluation_results.json', encoding='utf-8') as f:
    evaluation = json.load(f)
with open(ROOT / 'outputs/hyperparameter_results.json', encoding='utf-8') as f:
    comparison = json.load(f)
q_table = np.load(ROOT / 'models/q_table.npy')
print('Environment: FrozenLake-v1, 4x4, non-slippery')
print('Evaluation results:', evaluation)
print('Q-table shape:', q_table.shape)
print('Q-table:')
print(np.round(q_table, 3))
```
